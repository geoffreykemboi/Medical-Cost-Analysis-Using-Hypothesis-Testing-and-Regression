```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

X = df_cleaned.drop('mpg', axis=1)
y = df_cleaned['mpg']

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=254, test_size=0.2)
X_train.info()
```

```python
#instantiate model 
lr = LinearRegression()

lr.fit(X_train, y_train)
```

```python
y_preds = lr.predict(X_test)

### MAE, RMSE, MSE 

'y_preds & y_test'
```
